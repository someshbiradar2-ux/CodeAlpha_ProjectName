```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

file_path = "/content/drive/MyDrive/Netflix EDA Project/netflix_titles.csv"

df = pd.read_csv(file_path)

print("Dataset loaded successfully!")
```

```python
type_counts = df["type"].value_counts()

plt.figure(figsize=(8, 5))
plt.bar(type_counts.index, type_counts.values)

plt.title("Movies vs TV Shows on Netflix")
plt.xlabel("Content Type")
plt.ylabel("Number of Titles")

plt.show()
```

```python
year_counts = df["release_year"].value_counts().sort_index()

plt.figure(figsize=(12, 6))
plt.plot(year_counts.index, year_counts.values)

plt.title("Netflix Titles by Release Year")
plt.xlabel("Release Year")
plt.ylabel("Number of Titles")

plt.grid(True)
plt.show()
```

```python
country_counts = df["country"].value_counts().head(10)

plt.figure(figsize=(10, 6))
plt.bar(country_counts.index, country_counts.values)

plt.title("Top 10 Countries by Number of Netflix Titles")
plt.xlabel("Country")
plt.ylabel("Number of Titles")

plt.xticks(rotation=45, ha="right")
plt.tight_layout()

plt.show()
```

```python
rating_counts = df["rating"].value_counts().head(10)

plt.figure(figsize=(10, 6))
plt.bar(rating_counts.index, rating_counts.values)

plt.title("Top 10 Netflix Content Ratings")
plt.xlabel("Rating")
plt.ylabel("Number of Titles")

plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```

```python
genre_counts = df["listed_in"].value_counts().head(10)

plt.figure(figsize=(12, 7))
plt.barh(genre_counts.index, genre_counts.values)

plt.title("Top 10 Netflix Categories")
plt.xlabel("Number of Titles")
plt.ylabel("Category")

plt.gca().invert_yaxis()
plt.tight_layout()

plt.show()
```

```python
movie_data = df[df["type"] == "Movie"]

plt.figure(figsize=(10, 6))
plt.hist(movie_data["release_year"], bins=20)

plt.title("Distribution of Movie Release Years")
plt.xlabel("Release Year")
plt.ylabel("Number of Movies")

plt.tight_layout()
plt.show()
```

```python
tv_data = df[df["type"] == "TV Show"].copy()

tv_data["seasons"] = tv_data["duration"].str.extract(r"(\d+)").astype(float)

season_counts = tv_data["seasons"].value_counts().sort_index()

print(season_counts)
```

```python
plt.figure(figsize=(10, 6))
plt.bar(season_counts.index.astype(str), season_counts.values)

plt.title("Number of TV Shows by Seasons")
plt.xlabel("Number of Seasons")
plt.ylabel("Number of TV Shows")

plt.tight_layout()
plt.show()
```

```python
year_type = pd.crosstab(df["release_year"], df["type"])

plt.figure(figsize=(8, 12))
sns.heatmap(year_type, cmap="Blues", annot=False)

plt.title("Netflix Content by Release Year and Type")
plt.xlabel("Content Type")
plt.ylabel("Release Year")

plt.tight_layout()
plt.show()
```
